```python
# 집합
s1=set([1,2,34,4,5,6])
s2=set([4,5,6,7,8,9,43])
```

```python
print(s1&s2) #교집합
print(s1|s2) #합집합
print(s1-s2) #차집합
```

```python
s1.add(8) # 값 1개 추가
```

```python
s1
```

```python
s1.update([9,8,7]) # 중복은 무시함, 여러개 추가
```

```python
s1
```

```python
s1.remove(34)
```

```python
s1
```

```python
s1[0] # 집합은 인덱싱이 안됨, 중복무시
```
